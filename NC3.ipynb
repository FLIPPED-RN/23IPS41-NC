```python
from keras import preprocessing,models,utils
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Image
from keras.src.saving import load_model
```

```python
classes = ['футболка', 'брюки', 'свитер', 'платье', 'пальто', 'туфли', 'рубашка', 'кроссовки', 'сумка', 'ботинки']
```

```python
model=models.load_model('fashion_mnist_dense.keras')
model.summary()
```

```python
Image('sumka.jpg', width=400, height=400)
```

```python
img=preprocessing.image.load_img('sumka.jpg', target_size=(28,28), color_mode='grayscale')
x=preprocessing.image.img_to_array(img)
x=x.reshape((1,784))
x/=255
```

```python
prediction=model.predict(x)
prediction
```

```python
prediction=np.argmax(prediction)
print('номер класса: ', prediction)
print('название класса', classes[prediction])
```
